```python
# GPS Navigation System using Weighted Graph
# Dijkstra's Shortest Path Algorithm

import heapq

# City map with distances in kilometers
city_map = {
    "Home": {"School": 4, "Market": 2},
    "School": {"Home": 4, "Hospital": 5, "Park": 10},
    "Market": {"Home": 2, "Hospital": 8, "Park": 6},
    "Hospital": {"School": 5, "Market": 8, "Park": 3},
    "Park": {"School": 10, "Market": 6, "Hospital": 3}
}

def shortest_path(graph, start, destination):
    distances = {place: float("inf") for place in graph}
    previous = {place: None for place in graph}
    distances[start] = 0

    queue = [(0, start)]

    while queue:
        current_distance, current = heapq.heappop(queue)

        if current_distance > distances[current]:
            continue

        if current == destination:
            break

        for neighbor, distance in graph[current].items():
            new_distance = current_distance + distance

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                previous[neighbor] = current
                heapq.heappush(queue, (new_distance, neighbor))

    if distances[destination] == float("inf"):
        print("No route available.")
        return

    path = []
    current = destination

    while current is not None:
        path.append(current)
        current = previous[current]

    path.reverse()

    print("Shortest Route:", " -> ".join(path))
    print("Total Distance:", distances[destination], "km")


print("GPS Navigation System")
print("Available Locations:", ", ".join(city_map))

start = input("Enter starting location: ").title()
destination = input("Enter destination: ").title()

if start in city_map and destination in city_map:
    shortest_path(city_map, start, destination)
else:
    print("Invalid location! Please enter a listed location.")
```
