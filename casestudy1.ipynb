```python
# Case Study: E-commerce Product Sorting and Google Calendar

# E-commerce Product Sorting using Divide and Conquer
def merge_sort(products):
    if len(products) <= 1:
        return products

    mid = len(products) // 2
    left = merge_sort(products[:mid])
    right = merge_sort(products[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i][1] <= right[j][1]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    return result + left[i:] + right[j:]


# Google Calendar using Greedy Strategy
def schedule_events(events):
    events.sort(key=lambda x: x[1])
    selected = []
    last_end = 0

    for event in events:
        if event[0] >= last_end:
            selected.append(event)
            last_end = event[1]

    return selected


# Main Program
products = [
    ("Mobile", 15000),
    ("Laptop", 50000),
    ("Headphones", 2000),
    ("Tablet", 12000)
]

print("E-commerce Products Sorted by Price:")
for name, price in merge_sort(products):
    print(name, "- Rs.", price)

events = [
    (9, 10),
    (10, 12),
    (11, 13),
    (13, 14),
    (14, 16)
]

print("\nGoogle Calendar Scheduled Events:")
for start, end in schedule_events(events):
    print("Start:", start, "End:", end)
```
