```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    def push(self):
        n = int(input("How many elements do you want to push? "))
        a = []
        for i in range(n):
            x = int(input("Enter element: "))
            new = Node(x)
            new.next = self.top
            self.top = new
            a.append(x)
        print("Elements", *a, "pushed into stack")

    def pop(self):
        if not self.top:
            print("Stack is empty")
        else:
            print("Element", self.top.data, "popped")
            self.top = self.top.next

    def display(self):
        p = self.top
        print("Stack:", end=" ")
        while p:
            print(p.data, end=" ")
            p = p.next
        print()


class Queue:
    def __init__(self):
        self.front = self.rear = None

    def enqueue(self):
        n = int(input("How many elements do you want to enqueue? "))
        a = []
        for i in range(n):
            x = int(input("Enter element: "))
            new = Node(x)
            if not self.rear:
                self.front = self.rear = new
            else:
                self.rear.next = new
                self.rear = new
            a.append(x)
        print("Elements", *a, "added to queue")

    def dequeue(self):
        if not self.front:
            print("Queue is empty")
        else:
            print("Element", self.front.data, "removed")
            self.front = self.front.next
            if not self.front:
                self.rear = None

    def display(self):
        p = self.front
        print("Queue:", end=" ")
        while p:
            print(p.data, end=" ")
            p = p.next
        print()


s = Stack()
q = Queue()

while True:
    print("\n===== MAIN MENU =====")
    print("1. Stack Operations")
    print("2. Queue Operations")
    print("3. Exit")
    ch = int(input("Enter your choice: "))

    if ch == 1:
        while True:
            print("\n--- STACK OPERATIONS ---")
            print("1. Push\n2. Pop\n3. Display\n4. Exit")
            op = int(input("Enter operation: "))

            if op == 1: s.push()
            elif op == 2: s.pop()
            elif op == 3: s.display()
            elif op == 4: break
            else: print("Invalid choice")

    elif ch == 2:
        while True:
            print("\n--- QUEUE OPERATIONS ---")
            print("1. Enqueue\n2. Dequeue\n3. Display\n4. Exit")
            op = int(input("Enter operation: "))

            if op == 1: q.enqueue()
            elif op == 2: q.dequeue()
            elif op == 3: q.display()
            elif op == 4: break
            else: print("Invalid choice")

    elif ch == 3:
        print("Program ended.")
        break
    else:
        print("Invalid choice")
```
