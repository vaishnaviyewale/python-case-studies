```python
# Case Study: Social Network Adjacency Matrix

# List of users
users = ["Alice", "Bob", "Charlie", "David", "Emma", "Frank"]

n = len(users)

# Create an adjacency matrix
matrix = [[0 for j in range(n)] for i in range(n)]

# Function to add friendship
def add_friendship(user1, user2):
    if user1 in users and user2 in users:
        i = users.index(user1)
        j = users.index(user2)

        matrix[i][j] = 1
        matrix[j][i] = 1

        print(user1, "and", user2, "are now friends.")
    else:
        print("User not found!")


# Function to display adjacency matrix
def display_matrix():
    print("\nSocial Network Adjacency Matrix:")
    print("       ", end="")

    for user in users:
        print(user, end="   ")
    print()

    for i in range(n):
        print(users[i], end="   ")
        for j in range(n):
            print(matrix[i][j], end="       ")
        print()


# Function to display friends of a user
def show_friends(user):
    if user not in users:
        print("User not found!")
        return

    i = users.index(user)
    print("\nFriends of", user, ":")

    found = False
    for j in range(n):
        if matrix[i][j] == 1:
            print("-", users[j])
            found = True

    if not found:
        print("No friends found.")


# Function to count friends
def count_friends(user):
    if user not in users:
        print("User not found!")
        return

    i = users.index(user)
    count = sum(matrix[i])

    print(user, "has", count, "friends.")


# Function to remove friendship
def remove_friendship(user1, user2):
    if user1 in users and user2 in users:
        i = users.index(user1)
        j = users.index(user2)

        matrix[i][j] = 0
        matrix[j][i] = 0

        print("Friendship removed between", user1, "and", user2)
    else:
        print("User not found!")


# Add friendship connections
add_friendship("Alice", "Bob")
add_friendship("Alice", "Charlie")
add_friendship("Bob", "David")
add_friendship("Charlie", "Emma")
add_friendship("David", "Frank")
add_friendship("Emma", "Frank")

# Display matrix
display_matrix()

# Display friends
show_friends("Alice")
show_friends("Emma")

# Count friends
count_friends("Bob")

# Remove a friendship
remove_friendship("Alice", "Bob")

# Display updated matrix
display_matrix()
```
