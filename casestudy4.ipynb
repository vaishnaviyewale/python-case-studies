```python
# Case Study: Dictionary Application using Hash Tables

class Dictionary:
    def __init__(self, size):
        self.size = size
        self.table = [[] for i in range(size)]

    # Hash function
    def hash_function(self, word):
        return sum(ord(char) for char in word) % self.size

    # Add word and meaning
    def add_word(self, word, meaning):
        index = self.hash_function(word)

        for item in self.table[index]:
            if item[0] == word:
                item[1] = meaning
                print("Word updated successfully!")
                return

        self.table[index].append([word, meaning])
        print("Word added successfully!")

    # Search word
    def search_word(self, word):
        index = self.hash_function(word)

        for item in self.table[index]:
            if item[0] == word:
                print("Meaning of", word, ":", item[1])
                return

        print("Word not found!")

    # Delete word
    def delete_word(self, word):
        index = self.hash_function(word)

        for item in self.table[index]:
            if item[0] == word:
                self.table[index].remove(item)
                print("Word deleted successfully!")
                return

        print("Word not found!")

    # Display dictionary
    def display(self):
        print("\nDictionary Contents:")
        for i in range(self.size):
            print("Index", i, ":", self.table[i])


# Create dictionary
dictionary = Dictionary(10)

# Add words
dictionary.add_word("Apple", "A fruit")
dictionary.add_word("Book", "A collection of pages")
dictionary.add_word("Cat", "A domestic animal")
dictionary.add_word("Dog", "A loyal animal")
dictionary.add_word("Act", "To do something")

# Display dictionary
dictionary.display()

# Search words
print("\nSearching Words:")
dictionary.search_word("Apple")
dictionary.search_word("Dog")
dictionary.search_word("School")

# Update a word
print("\nUpdating Word:")
dictionary.add_word("Apple", "A sweet fruit")

# Display updated meaning
dictionary.search_word("Apple")

# Delete a word
print("\nDeleting Word:")
dictionary.delete_word("Cat")

# Display final dictionary
dictionary.display()
```
